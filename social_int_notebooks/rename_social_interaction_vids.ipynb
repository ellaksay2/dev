```python
import os
from pandas.io.sas.sas_constants import os_maker_length
```

```python
base_path = "C:/Users/esay/data/social_interaction"
# code_path = os.path.join(base_path, "Code", "forCanDong")
data_path = os.path.join(base_path, "SLEAP_raw", "videos")
```

```python
animal_ID = os.listdir(data_path)
```

```python
for animal in animal_ID:
    exp_day = os.listdir(os.path.join(data_path, animal))
    for day in exp_day:
        exp_session = os.listdir(os.path.join(data_path, animal, day))
        for session in exp_session:
            session_file = os.listdir(os.path.join(data_path, animal, day, session))
            for file in session_file:
                session_path = os.path.join(data_path, animal, day, session)
                extension = str.split(file, ".")[1]
                if "cam_23175601" in file:
                    os.rename(os.path.join(session_path, file), os.path.join(session_path, animal + "_pov_face." + extension))
                if "cam_23199174" in file:
                    os.rename(os.path.join(session_path, file), os.path.join(session_path, animal + "_pov_tunnel." + extension))
```
